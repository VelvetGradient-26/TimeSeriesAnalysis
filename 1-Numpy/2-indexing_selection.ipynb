```python
import numpy as np
```

```python
arr = np.arange(0, 11)
arr
```

```python
arr[3:]
```

```python
# General broadcasting
new_arr = arr + 1
print(new_arr)
```

```python
new_arr = arr * 10
print(new_arr)
```

```python
new_arr = arr / 2
print(new_arr)
```

```python
# copy an array via .copy() (coz numpy slices are referenced to the original array)
arr_copy = arr.copy()
print(arr_copy)
```

```python
# Two dimensional array indexing
np.random.seed(42)

array = np.random.rand(3,3)
print(array)


print(array[2][2])
print(array[1,1])
```

```python
# slicing
array[:2, :1]
```

```python
# conditional selection
array = np.arange(1, 11)

array > 4
```

```python
bool_arr = arr > 4
bool_arr
```

```python
arr[bool_arr]
```
